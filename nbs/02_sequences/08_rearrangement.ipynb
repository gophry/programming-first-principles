#### Swap

```python
#| exports
def swap(a, i, j):
    a[i], a[j] = a[j], a[i]
```

```python
arr = [1, 2, 3, 4, 5]
f'{arr = }'
swap(arr, 0, 1)
f'{arr = }'
```

#### Reverse

```python
#| exports
def reverse(a,f,l):
  """
  Reverses elements in the half-open range [f, l)
  """
  l -= 1
  while f < l:
    a[f], a[l] = a[l], a[f]
    f += 1
    l -= 1
```

```python
arr = [1,2,3,4,5,6,7,8]
f'Before: {arr}'
reverse(arr,1, 6)
f'After:  {arr}'
```

#### Rotate

```python
#| exports
def rotate(a, first, middle, last):
    reverse(a, first, middle)
    reverse(a, middle, last)
    reverse(a, first, last)
    return first + (last - middle)
```

```python
arr = [0, 1, 2, 3, 4, 5]
f'Before: {arr}'
f'{rotate(arr, 0, 2, 6) = }'
f'After:  {arr}'
```

```python
arr = [0, 1, 2, 3, 4, 5]
f'{rotate(arr, 0, 0, 6) = }'
f'{rotate(arr, 0, 6, 6) = }'
f'{arr = }'
```

#### Slide

```python
#| exports
def slide(a, f, l, p):
    if p < f:
        return p, rotate(a, p, f, l)
    if l < p:
        return rotate(a, f, l, p), p
    return f, l
```

```python
a = [0, 1, 2, 3, 4, 5]

nf, nl = slide(a, 0, 2, 6)

f'{nf = }, {nl = }'
f'{a = }'
f'{slide(a, nf, nl, 0) = }'
f'{a = }'
f'{slide(a, 3, 5, 0) = }'
f'{a = }'
```
