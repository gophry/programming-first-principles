## Mutability
* content inside sequences can changed post creation.
* `__setitem__`
* `___delitem__`



#### MutableSequence - `__setitem__`

```python
a = my_array[:]

f'Before: {a = }'
a.__setitem__(4,350)
f'After:  {a = }'

from typing import MutableSequence
f'{isinstance(my_array,MutableSequence) = }'
```

#### Subscript assignment

```python
a = my_array[:]
f'{a is my_array = }'
f'{a == my_array = }'

f'Before: {a = }'
a[4] = 350
f'After:  {a = }'
f'{a == my_array = }'
```

#### `IndexError`

```python
a = my_array[:]

f'Before: {a = }'
print()
try:
  a.__setitem__(7,350)
except IndexError as e:
  f'a.__setitem__(7,350): {e}'
```

#### Python Object View
* `id(a[i])` gives object’s heap address.
* temporary boxed Python objects obtained by iteration
* Address of the boxed int, not the address of the array slot.


```python
a = my_array[:]
f'Address of {id(a) = })'
f'Before Modification Address'
for j in a:
  f'{j = }, {id(j)}'

a[4] = 350
print()
f'After Modification Address'
f'Address of a ,id(a): {id(a)})'
for j in a:
  f'{j = }, {id(j)}'

print('not looking sequential')
```

#### Buffer View
* contiguous native memory exposed by the Buffer Protocol
* Address of the slot

```python
def slot_addr(i, base, itemsize):
    return base + i * itemsize

a = my_array[:]
base, n = a.buffer_info()
itemsize = a.itemsize

f'Before Modification slot'
for i, j in enumerate(a):
    f'{i = }, {j = }, {slot_addr(i, base, itemsize)}'

a[4] = 350

f'After Modification slot'
for i, j in enumerate(a):
    f'{i = }, {j = }, {slot_addr(i, base, itemsize)}'
```
