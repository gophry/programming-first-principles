# Sequences - Tuple, List


## Tuple

#### tuple
* `a,` or `(a,)`
* `a, b, c` or `(a, b, c)`
* `()` — empty
* `tuple()` / `tuple(iterable)`

```python
t1 = 501,
print(f'a,      - {t1}, {type(t1)}')
t2 = (501,)
print(f'(a,)    - {t2}, {type(t2)}')
t3 = 501,502,503
print(f'a,b,c   - {t3}, {type(t3)}')
t4 = (501,502,503)
print(f'(a,b,c) - {t4}, {type(t4)}')
t5 = ()
print(f'()      - {t5}, {type(t5)}')
```

#### Immutability

```python
t = (3, 4)
try:
  t[0] = 9
except TypeError as e:
  f'{e =}'
```

#### Tuple as Record

```python
point = (3, 4)          # (x, y)
employee = ("Yogesh", "Thane", 8)

f'{point = }'
f'{employee = }'
```

#### Tuple Equality

```python
a = ()
b = ()

f'{a}, {b}, {a is b = }'

a = (1,2)
b = (1,2)

f'{a = }, {b = }, {a is b = }, {a == b = }'
```

#### Concatenation / Repetition

```python
t1 = (1,2)
t2 = (3,4)

f'{t1 + t2 =}'
f'{t1.__add__(t2) =}'

f'{t1 * 3 = }'
f'{t1.__mul__(3) =}'
```

#### Nested Tuples

```python
t = (1, (2, 3), (4, (5, 6)))
t[1]

f'{t[1][0] = }'
f'{t[2][1][1] = }'

a, (b, c), (d, (e, f)) = t     # unpacking mirrors the nesting shape
# a, b, c, d, e, f

f'{a =}, {b =}, {c =}, {d =}, {e =}, {f =}'
```

## Tuple Packing/Unpacking

* Slicing using the fundmentals could extract a sub sequence.
* `x,y = my_array`

#### Basic Unpacking
* Assign Values from elements at positions onto variables.

```python
arr = [1,2]

f'arr: {arr} '

x1,x2 = arr

f'x1: {x1}'
f'x2: {x2}'
f'arr: {arr} '
```

#### Too Many Values to Unpack
* Assign Values from elements at positions onto variables.

```python
my_array = array("i",[501, 502, 503, 504, 505])
try:
  first,last = my_array
except ValueError as e:
  f'{e = }'
```

#### Star Expression - Basic
* `*` expression unpacks into a `list` sequence.

```python
first,*mid,last = my_array
f'first: {first}'
f'{mid = }, {type(mid) = }  '
f'last: {last =}'
```

#### Star Expression - Usage


```python
*lead, end = my_array
f'{lead =}, {end =}'
first, *mid, last = my_array
f'{first = }, {mid = }, {last =}'
z1, *tail = my_array
f'first = {first}, tail = {tail}'

*alone, = my_array
```

#### Not enough values to unpack

```python
try:
  v1, v2, v3, v4, v5, v6 = my_array
except ValueError as e:
  f'{e =}'
```

#### Discard, CatchAll

```python
_, second,*end  = my_array
f'Discarded: `_`: {_} '
f'second: {second}'

first, second,*_  = my_array
print(*_)
```

#### Enumerate

```python
f'{'__iter__' in dir(enumerate(my_array)) = }'
f'{'__next__' in dir(enumerate(my_array)) = }'
f'{'__getitem__' in dir(enumerate(my_array)) = }'
for item in enumerate(my_array):
  f'{item = }'
```

```python
for i, item in enumerate(my_array):
  f'{i = },{item = }'
```

#### Zip

```python
other_array = array("i", [10, 20, 30, 40, 50])
f'{"__iter__" in dir(zip(my_array, other_array)) = }'
f'{"__next__" in dir(zip(my_array, other_array)) = }'
f'{"__getitem__" in dir(zip(my_array, other_array)) = }'
```

```python
for item in zip(my_array, other_array):
    f'{item = }'
```

## NamedTuple
* Named Field Access with Immutability
* Perfect for Records

#### Named Field Access

```python
Point = namedtuple('Point', ['x', 'y'])
p = Point(3, 4)

f'{p.x = }, {p.y = }'

f'{p[0] = }, {p[1] = }'
```

## List
* Drops the typecode and provides homegenity in data conceptually.
* Mutable like array.
* `[]`, `list()`, `list(iterable)`

#### List: Dropping the Typecode

```python
a = array('i', [1, 2, 3])

try:
  a.append(4.5)
except TypeError as e:
  f'{a = }, {e = }'
```

```python
l = [1, 2, 3]
l.append(4.5)
l.append("five")

f'{l = }'

scores = [88, 92, 76, 95]     # intended usage: homogeneous
scores.append("N/A")          # language permits this — nothing stops it
f'{scores = }'
```

#### Size Array, Tuple, List

#### List Comprehension

```python
a = [1,2,3,4,5]
b = []
for x in a:
  if x % 2 == 0:
    b.append(x)
f'For Loop {b = }'

c = [y for y in a if y%2 ==0]
f'List Comprehension: {c = }'
```
