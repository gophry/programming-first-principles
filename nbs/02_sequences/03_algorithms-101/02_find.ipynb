#### Find Element
*  `==` equality operator

```python
#| exports
def find(arr, target):
    index = 0
    for value in arr:          # iterator walk
        if value == target:    # equality
            return index       # found
        index = index + 1
    return -1                  # not found
```

```python
arr = array('i', [40, 11, 73, 8, 25, 91, 16, 4])
f'{25 == 25 = } '
f'{25 == 91 = } '

f'{arr[4] == 25 = }'
f'{arr[5] == 91 = }'
f'{find(arr,25) = }'
f'{find(arr,99) = }'
```

#### find_if

```python
#| exports
def find_if(a, first, last, pred):
    for i in range(first, last):
        if pred(a[i]):
            return i
    return last
```

```python
# from programming_first_principles.predicates import is_even
scores_fi = [5, 3, 40, 10, 20, 60, 75, 80]
f'{find_if(scores_fi, 0, len(scores_fi), is_even) = }'
```

```python
find_if_measure = Cost(repeats=1)
_ = find_if_measure.run(find_if, scores_fi, 0, len(arr), lambda x: x == 10,
         name='find_if', shape='found at 4', operations=('predicates','ms'))
_ = find_if_measure.run(find_if, scores_fi, 0, len(arr), lambda x: x == 90,
         name='find_if', shape='absent', operations=('predicates','ms'))
find_if_measure.frame
```

#### find_if_not

```python
#| exports
def find_if_not(a, first, last, pred):
    for i in range(first, last):
        if not pred(a[i]):
            return i
    return last
```

```python
scores_fin = [5, 3, 40, 10, 20, 60, 75, 80]
f'{find_if_not(scores_fin, 0, len(scores_fin), is_even) = }'
```

```python
fin_measure = Cost(repeats=1)
_ = fin_measure.run(find_if_not, [40, 8, 16, 4, 11, 25], 0, 6, is_even,
         name='find_if_not', shape='fails at 4', operations=('predicates','ms'))
_ = fin_measure.run(find_if_not, [40, 8, 16, 4, 2, 10], 0, 6, is_even,
         name='find_if_not', shape='all pass', operations=('predicates','ms'))
fin_measure.frame
```

#### find_mismatch

```python
#| exports
def find_mismatch(a, f0, l0, b, f1, l1, r):
    while f0 != l0 and f1 != l1 and r(a[f0], b[f1]):
        f0 += 1
        f1 += 1
    return f0, f1
```

```python
# a = [1, 2, 9, 4]
# b = [1, 2, 4]
a = [1, 2, 9, 4]
b = [7, 7, 1, 2, 4, 5]

f'{find_mismatch(a, 0, len(a), b, 2, 5, operator.eq) = }'  # (2, 2)
```

```python
mm = Cost(repeats=1)
_ = mm.run(find_mismatch, [1, 2, 9, 4], 0, 4, [1, 2, 4, 5], 0, 4, operator.eq,
           name='find_mismatch', shape='mismatch at 2', operations=('predicates','ms'))
_ = mm.run(find_mismatch, [1, 2, 3, 4], 0, 4, [1, 2, 3, 4], 0, 4, operator.eq,
           name='find_mismatch', shape='equal', operations=('predicates','ms'))
mm.frame
```

#### find_adjacent_mismatch

```python
#| exports
def find_adjacent_mismatch(a, first, last, pred):
    if first == last:
        return last
    x = a[first]
    first += 1
    while first != last and pred(x, a[first]):
        x = a[first]
        first += 1
    return first
```

```python
a = [1, 2, 2, 0]
f'{find_adjacent_mismatch(a, 0, len(a), lambda x, y: x <= y) = }' 
```

```python
adj = Cost(repeats=1)
_ = adj.run(find_adjacent_mismatch, [2, 1, 3, 4], 0, 4, lambda x, y: x <= y,
            name='find_adjacent_mismatch', shape='breaks at 1', operations=('predicates','ms'))
_ = adj.run(find_adjacent_mismatch, [1, 2, 3, 4], 0, 4, lambda x, y: x <= y,
            name='find_adjacent_mismatch', shape='ordered', operations=('predicates','ms'))
adj.frame
```

#### Count

```python
#| exports
def count_if(a, first, last, pred):
    count = 0 
    for i in range(first, last):
        if pred(a[i]):
            count += 1
    return count
```

```python
t = array("i",[1,2,1,2,3,3,5,6,2])
f'{count_if(t, 0, len(t), partial(operator.eq, 2)) = }'
f'{t.count(2) = }'
```

```python
cnt = Cost(repeats=1)
t = [1, 2, 1, 2, 3, 3, 5, 6, 2]
_ = cnt.run(count_if, t, 0, len(t), partial(operator.eq, 2),
            name='count_if', shape='three hits', operations=('predicates','ms'))    
_ = cnt.run(count_if, t, 0, len(t), partial(operator.eq, 9),    
            name='count_if', shape='none', operations=('predicates','ms'))
cnt.frame
```

## Quantifier
* A quantifier says how many elements of a range satisfy a predicate, and the function returns only that yes or no.

#### all_of

```python
#| exports
def all_of(a, first, last, pred): 
    return find_if_not(a, first, last, pred) == last    
```

```python
scores_ao = [5, 3, 40, 10, 20, 60, 75, 80]
f'{all_of(scores_ao, 0, len(scores_ao), is_even) = }'
```

```python
ao = Cost(repeats=1)
_ = ao.run(all_of, [5, 3, 40, 10], 0, 4, is_even,
           name='all_of', shape='fails at 0', operations=('predicates','ms'))   
_ = ao.run(all_of, [40, 8, 16, 4], 0, 4, is_even,
           name='all_of', shape='all pass', operations=('predicates','ms'))
ao.frame
```

#### any_of

```python
#| exports
def any_of(a, first, last, pred):
    return find_if(a, first, last, pred) != last
```

```python
scores_any = [5, 3, 40, 10, 20, 60, 75, 80]
f'{any_of(scores_any, 0, len(scores_any), is_even) = }'
```

```python
any_m = Cost(repeats=1)
_ = any_m.run(any_of, [5, 3, 40, 10], 0, 4, is_even,
              name='any_of', shape='found at 2', operations=('predicates','ms'))
_ = any_m.run(any_of, [5, 3, 7, 9], 0, 4, is_even,
              name='any_of', shape='absent', operations=('predicates','ms'))
any_m.frame
```

#### none_of

```python
#| exports
def none_of(a, first, last, pred):
    return find_if(a, first, last, pred) == last
```

```python
scores_none = [5, 3, 40, 10, 20, 60, 75, 80]
f'{none_of(scores_none, 0, len(scores_none), is_even) = }'
```

```python
none_m = Cost(repeats=1)
_ = none_m.run(none_of, [5, 3, 40, 10], 0, 4, is_even,
               name='none_of', shape='found at 2', operations=('predicates','ms'))
_ = none_m.run(none_of, [5, 3, 7, 9], 0, 4, is_even,
               name='none_of', shape='none', operations=('predicates','ms'))
none_m.frame
```
