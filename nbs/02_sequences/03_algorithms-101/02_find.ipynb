#### Find Element
*  `==` equality operator

```python
#| exports
def find(arr, target):
    index = 0
    for value in arr:          # iterator walk
        if value == target:    # equality
            return index       # found
        index = index + 1
    return -1                  # not found
```

```python
arr = array('i', [40, 11, 73, 8, 25, 91, 16, 4])
f'{25 == 25 = } '
f'{25 == 91 = } '

f'{arr[4] == 25 = }'
f'{arr[5] == 91 = }'
f'{find(arr,25) = }'
f'{find(arr,99) = }'
```

#### find_if

```python
def is_even(x):
    return x % 2 == 0
```

```python
#| exports
def find_if(a, first, last, pred):
    for i in range(first, last):
        if pred(a[i]):
            return i
    return last
```

```python
# from programming_first_principles.predicates import is_even
scores_c = [5, 3, 40, 10, 20, 60, 75, 80]
f'{find_if(scores_c, 0, len(scores_c), is_even) = }'
```

#### find_if_not

```python
#| exports
def find_if_not(a, first, last, pred):
    for i in range(first, last):
        if not pred(a[i]):
            return i
    return last
```

```python
scores_c = [5, 3, 40, 10, 20, 60, 75, 80]
f'{find_if_not(scores_c, 0, len(scores_c), is_even) = }'
```

#### find_mismatch

```python
#| exports
def find_mismatch(a, f0, l0, b, f1, l1, r):
    while f0 != l0 and f1 != l1 and r(a[f0], b[f1]):
        f0 += 1
        f1 += 1
    return f0, f1
```

```python
# a = [1, 2, 9, 4]
# b = [1, 2, 4]
a = [1, 2, 9, 4]
b = [7, 7, 1, 2, 4, 5]

f'{find_mismatch(a, 0, len(a), b, 2, 5, operator.eq) = }'  # (2, 2)
```

#### find_adjacent_mismatch

```python
#| exports
def find_adjacent_mismatch(a, first, last, pred):
    if first == last:
        return last
    x = a[first]
    first += 1
    while first != last and pred(x, a[first]):
        x = a[first]
        first += 1
    return first
```

```python
a = [1, 2, 2, 0]
f'{find_adjacent_mismatch(a, 0, len(a), lambda x, y: x <= y) = }' 
```

#### Count

```python
#| exports
def count_if(a, first, last, pred):
    count = 0 
    for i in range(first, last):
        if pred(a[i]):
            count += 1
    return count
```

## Quantifier
* A quantifier says how many elements of a range satisfy a predicate, and the function returns only that yes or no.

#### all_of

```python
#| exports
def all_of(a, first, last, pred): 
    return find_if_not(a, first, last, pred) == last    
```

```python
scores_c = [5, 3, 40, 10, 20, 60, 75, 80]
f'{all_of(scores_c, 0, len(scores_c), is_even) = }'
```

#### any_of

```python
#| exports
def any_of(a, first, last, pred):
    return find_if(a, first, last, pred) != last
```

```python
scores_c = [5, 3, 40, 10, 20, 60, 75, 80]
f'{any_of(scores_c, 0, len(scores_c), is_even) = }'
```

#### none_of

```python
#| exports
def none_of(a, first, last, pred):
    return find_if(a, first, last, pred) == last
```

```python
scores_c = [5, 3, 40, 10, 20, 60, 75, 80]
f'{none_of(scores_c, 0, len(scores_c), is_even) = }'
```
