#### Swap

```python
#| exports
def swap(a, i, j):
    a[i], a[j] = a[j], a[i]
```

```python
arr = [1, 2, 3, 4, 5]
f'{arr = }'
swap(arr, 0, 1)
f'{arr = }'
```

#### Reverse

```python
#| exports
def reverse(seq,first,last):
  """
  Reverses elements in the half-open range [f, l) using swap and iteration
  """
  last -= 1
  while first < last:
    seq[first], seq[last] = seq[last], seq[first]
    first += 1
    last -= 1
```

```python
arr = [1,2,3,4,5,6,7,8]
f'Before: {arr}'
reverse(arr,1, 6)
f'After:  {arr}'
```

```python
rev_swap = Cost(repeats=1)
_ = rev_swap.run(reverse, [1, 2, 3, 4, 5, 6, 7, 8], 0, 8,
                 name='reverse', shape='even', operations=('swaps','ms'))
_ = rev_swap.run(reverse, [1, 2, 3, 4, 5], 0, 5,
                 name='reverse', shape='odd', operations=('swaps','ms'))
rev_swap.frame
```

```python
#| exports
def reverse_py(seq, first, last):
    """Reverses elements in the half-open range [first, last) using slicing"""
    seq[first:last] = seq[first:last][::-1]
```

```python
arr1 = [1,2,3,4,5,6,7,8]
f'Before: {arr1}'
reverse_py(arr1,1, 6)
f'After:  {arr1}'
```

```python
rev_slice = Cost(repeats=1)
_ = rev_slice.run(reverse, [1, 2, 3, 4, 5, 6, 7, 8], 0, 8,
                  name='reverse', shape='even', operations=('swaps','ms'))
_ = rev_slice.run(reverse, [1, 2, 3, 4, 5], 0, 5,
                  name='reverse', shape='odd', operations=('swaps','ms'))
rev_slice.frame
```

#### Rotate

```python
#| exports
def rotate(a, first, middle, last):
    reverse(a, first, middle)
    reverse(a, middle, last)
    reverse(a, first, last)
    return first + (last - middle)
```

```python
arr = [0, 1, 2, 3, 4, 5]
f'Before: {arr}'
f'{rotate(arr, 0, 2, 6) = }'
f'After:  {arr}'
```

```python
rot = Cost(repeats=1)
_ = rot.run(rotate, [0, 1, 2, 3, 4, 5], 0, 2, 6,
            name='rotate', shape='middle at 2', operations=('swaps','ms'))
_ = rot.run(rotate, [0, 1, 2, 3, 4, 5], 0, 0, 6,
            name='rotate', shape='middle at start', operations=('swaps','ms'))
_ = rot.run(rotate, [0, 1, 2, 3, 4, 5], 0, 6, 6,
            name='rotate', shape='middle at end', operations=('swaps','ms'))
rot.frame
```

```python
#| exports
def rotate_py(a, first, middle, last):
    reverse_py(a, first, middle)
    reverse_py(a, middle, last)
    reverse_py(a, first, last)
    return first + (last - middle)
```

```python
arr = [0, 1, 2, 3, 4, 5]
f'{rotate_py(arr, 0, 0, 6) = }'
f'{rotate_py(arr, 0, 6, 6) = }'
f'{arr = }'
```

#### Slide

```python
#| exports
def slide(a, f, l, p):
    if p < f:
        return p, rotate(a, p, f, l)
    if l < p:
        return rotate(a, f, l, p), p
    return f, l
```

```python
a = [0, 1, 2, 3, 4, 5]

nf, nl = slide(a, 0, 2, 6)

f'{nf = }, {nl = }'
f'{a = }'
f'{slide(a, nf, nl, 0) = }'
f'{a = }'
f'{slide(a, 3, 5, 0) = }'
f'{a = }'
```

```python
sld = Cost(repeats=1)
_ = sld.run(slide, [0, 1, 2, 3, 4, 5], 0, 2, 6,
            name='slide', shape='p after range', operations=('swaps','ms'))
_ = sld.run(slide, [0, 1, 2, 3, 4, 5], 3, 5, 0,
            name='slide', shape='p before range', operations=('swaps','ms'))
_ = sld.run(slide, [0, 1, 2, 3, 4, 5], 0, 2, 1,
            name='slide', shape='p inside range', operations=('swaps','ms'))
sld.frame
```
