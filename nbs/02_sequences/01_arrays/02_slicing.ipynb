## Slicing
1. Sub Sequence
2. slice()
3. Subscripts

#### Slicing - start, stop

```python
f'Slicing an array, gives you a new array'
start, stop = 1, 3
out = array('i')

i = start
while i < stop:
    try:
        out.append(my_array[i])
    except IndexError:
        break
    i += 1
print(out)
```

#### Slicing - step

```python
f'Slicing an array, gives you a new array'
start, stop, step = 0, 5, 2
out = array('i')

i = start
while i < stop:
    try:
        out.append(my_array[i])
    except IndexError:
        break
    i += step
print(out)
```

#### Comma-separated subscripts
* `slice(start,stop,step)`
* `start` as None assumes index as 0
* `stop` as None assumes index len(my_array) -1
* `step` as None assumes 1

```python
f'{my_array[slice(None,None,None)] = }'
```

```python
f'{my_array[slice(1, 3)] = }'
```

```python
f'{my_array[slice(0,5,2)]= }'
```

```python
f'{my_array[slice(0,None,3)] = }'
```

#### Slice Subcript
* `arr[start,stop,step]`
* Missing slice items are always filled in with None.

```python
f'myarray[start:stop]: {my_array[1:3]}'
```

```python
f'myarray[start:stop:step]: {my_array[0:5:2]}'
```

```python
f'myarray[:stop:step]: {my_array[:5:2]}'
```

#### Slice Negative Subscript

```python
f'Everything except last: myarray[:-1]: {my_array[:-1]}'
```
