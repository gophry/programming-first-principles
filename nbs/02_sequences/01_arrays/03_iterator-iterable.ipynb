## Iterator and Iterable
* **Iterable** → `__iter__` (or `__getitem__` fallback)
* **Iterator** → `__next__`


* PEP 234
  * The two methods correspond to two distinct protocols:
    * An object can be iterated over with for if it implements __iter__() or __getitem__().
    * An object can function as an iterator if it implements next().

  * Container-like objects usually support protocol 1. Iterators are currently required to support both protocols. The semantics of iteration come only from protocol 2; protocol 1 is present to make iterators behave like sequences; in particular so that code receiving an iterator can use a for-loop over the iterator.

#### `while` Loop
* Base Address + Offset → While loop

```python
base, n = my_array.buffer_info()
size = my_array.itemsize

i = 0
while i < n:
    f'{c_int.from_address(base + i * size).value}'
    i += 1
```

#### Iterable
* *PEP234(protocol 1)* - An object can be iterated over with for if it implements `__iter__()` or `__getitem__()`.
* Iterable - `__iter__` is present on array.
* protocol 1 only — `for` works, next(obj) fails.


```python
f'{'__iter__' in dir(my_array) = }'
```

```python
f'{'__getitem__' in dir(my_array) = }'
```

#### Iterator
* `Iterator` - `__next__`

* `iter(my_array)` makes it into a protocol 2 object on which next() can be used.
* useful if an api wants to consume a iterator. (iter() converts protocol1 to iterator)
* `StopIteration`

```python
try:
  next(my_array)
except  TypeError as e:
  f'next(my_array): {e}'
f'{'__next__' in dir(my_array) = }'
```

```python
i = 0
iter_my_array = iter(my_array)
while True:
    try:
        f'{next(iter_my_array) = }'
    except StopIteration:
        break
    i += 1
```

```python
try:
    next(iter_my_array)
except StopIteration as e:
    f'next(iter_my_array): {e.__class__}'
```

#### Iterator  ⊂  Iterable
* Iterator  ⊂  Iterable
* Every iterator is an iterable
* Not every iterable is an iterator
  * containers have `__iter__` but no `__next__`

```python
iter_my_array = iter(my_array)
f'my_array is a Iterable: {isinstance(my_array, Iterable) = }'
f'my_array is a Iterator: {isinstance(my_array, Iterator) = }'
```

```python
f'iter_my_array is a Iterable: {isinstance(iter_my_array, Iterable) = }'
f'iter_my_array is a Iterator: {isinstance(iter_my_array, Iterator) = }'
```

#### iterator exhaustion

* A container object (such as a list) produces a fresh new iterator each time you pass it to the iter() function or use it in a for loop.

* Attempting this with an iterator will just return the same exhausted iterator object used in the previous iteration pass, making it appear like an empty container.


```python
f'Iterable - First round'
for item in my_array:
  f'{item = }'
```

```python
f'Iterable - Second round'
for item in my_array:
  f'{item = }'
```

```python

iter_my_array = iter(my_array)
f'{type(iter_my_array) = }'

f'Iterator - First round'
for item in iter_my_array:
  f'{item = }'
```

```python
f'Iterator - Second round'
try:
  f'{next(iter_my_array) = }'
except StopIteration as e:
  f'next(iter_my_array): {repr(e)}'
```
