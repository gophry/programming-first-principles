## Subscription
* The subscription syntax is usually used for selecting an element from a collection.
* In the subscription syntax, the object being subscribed - a primary - is followed by a subscript in square brackets.
* Subscript is called a `index` in sequences.

* Subcription Protocol
  * **Sized** → `__len__` - How Many elements?
  * **Sequence** → `__getitem__` - What is at index `i` **and** `__len__`  
  * (ordered, efficient access by integer index)

> methods `__methodname__` which start and end with are called dunder, are special methods.

```python
help(my_array.__len__)
help(my_array.__getitem__)
```

#### Sequence and Sized
* `__getitem__` and `__len__`
* `my_array.__getitem__(i)` replaces `ctypes.c_int.from_address(base + i * itemsize).value`

```python
f'4th element: {my_array.__getitem__(3) = }'
print()
f'{my_array.__len__() = }'
f'{len(my_array) = }'
print()
try:
  f'6th element{my_array.__getitem__(5)}'
except IndexError as e:
  f'6th element: my_array.__getitem__(5): {e}'
```

```python
i = 0
while True:
    try:
        f'my_array.__getitem__({i}) = {my_array.__getitem__(i)}'
    except IndexError:
        break
    i += 1
```

#### Subcript/Index
* `[]` is a syntatic sugar for `__getitem__`
* Object is Subscriptable if [] works on it.
* A numeric value that represents the position of an element in a sequence.
* `arr[subscript]` or `arr.__getitem__(subscript)`, where subscript (also called index for sequences) should be an integer or a slice object



```python
f'4th element: {my_array[3] = }'
try:
  f'6th element{my_array[5]}'
except IndexError as e:
  f'6th element: my_array[5]: {e}'
```

#### Negative Subscript
* `arr[-1]` will get the last element.
* its equivalent to arr[len(arr)-1], where `n` is length of the array.

```python
f'Last element: {my_array[-1] = }'
f'Last element: {my_array[len(my_array) - 1] = }'
```
