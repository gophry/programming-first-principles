# Packing/Unpacking
* Slicing using the fundmentals could extract a sub sequence.
* `x,y = my_array`

#### Basic Unpacking
* Assign Values from elements at positions onto variables.

```python
arr = (1,2)

f'arr: {arr} '

x1,x2 = arr

f'x1: {x1}'
f'x2: {x2}'
f'arr: {arr} '
```

#### Too Many Values to Unpack
* Assign Values from elements at positions onto variables.

```python
my_array = array("i",[501, 502, 503, 504, 505])
try:
  first,last = my_array
except ValueError as e:
  f'{e = }'
```

#### Star Expression - Basic
* `*` expression unpacks into a `list` sequence.

```python
first,*mid,last = my_array
f'first: {first}'
f'{mid = }, {type(mid) = }  '
f'last: {last =}'
```

#### Star Expression - Usage


```python
*lead, end = my_array
f'{lead =}, {end =}'
first, *mid, last = my_array
f'{first = }, {mid = }, {last =}'
z1, *tail = my_array
f'first = {first}, tail = {tail}'

*alone, = my_array
```

#### Not enough values to unpack

```python
try:
  v1, v2, v3, v4, v5, v6 = my_array
except ValueError as e:
  f'{e =}'
```

#### Discard, CatchAll

```python
_, second,*end  = my_array
f'Discarded: `_`: {_} '
f'second: {second}'

first, second,*_  = my_array
print(*_)
```

#### Enumerate

```python
f'{'__iter__' in dir(enumerate(my_array)) = }'
f'{'__next__' in dir(enumerate(my_array)) = }'
f'{'__getitem__' in dir(enumerate(my_array)) = }'
for item in enumerate(my_array):
  f'{item = }'
```

```python
for i, item in enumerate(my_array):
  f'{i = },{item = }'
```

#### Zip

```python
other_array = array("i", [10, 20, 30, 40, 50])
f'{"__iter__" in dir(zip(my_array, other_array)) = }'
f'{"__next__" in dir(zip(my_array, other_array)) = }'
f'{"__getitem__" in dir(zip(my_array, other_array)) = }'
```

```python
for item in zip(my_array, other_array):
    f'{item = }'
```
