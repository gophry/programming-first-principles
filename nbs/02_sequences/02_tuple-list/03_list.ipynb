# List
* Drops the typecode and provides homegenity in data conceptually.
* Mutable like array.
* `[]`, `list()`, `list(iterable)`

#### List: Dropping the Typecode

```python
a = array('i', [1, 2, 3])

try:
  a.append(4.5)
except TypeError as e:
  f'{a = }, {e = }'
```

```python
l = [1, 2, 3]
l.append(4.5)
l.append("five")

f'{l = }'

scores = [88, 92, 76, 95]     # intended usage: homogeneous
scores.append("N/A")          # language permits this — nothing stops it
f'{scores = }'
```

#### List Comprehension

```python
a = [1,2,3,4,5]
b = []
for x in a:
  if x % 2 == 0:
    b.append(x)
f'For Loop {b = }'
```

```python
c = [y for y in a if y%2 ==0]
f'List Comprehension: {c = }'
```
