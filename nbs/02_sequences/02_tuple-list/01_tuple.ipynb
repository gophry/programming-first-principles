# Tuple

#### tuple
* `a,` or `(a,)`
* `a, b, c` or `(a, b, c)`
* `()` — empty
* `tuple()` / `tuple(iterable)`

```python
t1 = 501,
print(f'a,      - {t1}, {type(t1)}')
t2 = (501,)
print(f'(a,)    - {t2}, {type(t2)}')
t3 = 501,502,503
print(f'a,b,c   - {t3}, {type(t3)}')
t4 = (501,502,503)
print(f'(a,b,c) - {t4}, {type(t4)}')
t5 = ()
print(f'()      - {t5}, {type(t5)}')
```

#### Immutability

```python
t = (3, 4)
try:
  t[0] = 9
except TypeError as e:
  f'{e =}'
```

#### Tuple as Record

```python
point = (3, 4)          # (x, y)
employee = ("Yogesh", "Thane", 8)

f'{point = }'
f'{employee = }'
```

#### Tuple Equality

```python
a = ()
b = ()

f'{a}, {b}, {a is b = }'

a = (1,2)
b = (1,2)

f'{a = }, {b = }, {a is b = }, {a == b = }'
```

#### Concatenation / Repetition

```python
t1 = (1,2)
t2 = (3,4)

f'{t1 + t2 =}'
f'{t1.__add__(t2) =}'

f'{t1 * 3 = }'
f'{t1.__mul__(3) =}'
```

#### Nested Tuples

```python
t = (1, (2, 3), (4, (5, 6)))
t[1]

f'{t[1][0] = }'
f'{t[2][1][1] = }'

a, (b, c), (d, (e, f)) = t     # unpacking mirrors the nesting shape
# a, b, c, d, e, f

f'{a =}, {b =}, {c =}, {d =}, {e =}, {f =}'
```

## NamedTuple
* Named Field Access with Immutability
* Perfect for Records

#### Named Field Access

```python
Point = namedtuple('Point', ['x', 'y'])
p = Point(3, 4)

f'{p.x = }, {p.y = }'

f'{p[0] = }, {p[1] = }'
```
