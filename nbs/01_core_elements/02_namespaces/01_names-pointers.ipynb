# Names, Pointers

#### Chained Assignemnt

```python
a = 350
b = a

x = y = 350
f'{id(a) = }, {id(b) = }, {a is b = }, {a == b = }'
# assignment binds a name to an existing object — it doesn't copy
f'{id(x) = }, {id(y) = }, {x is y = }, {x == y = }'
```

### Values have type

```python
a = 5
a = "now a string"
f'{a = }, {type(a) = }'
```

#### Rebinding

```python
a = "hi"
b = a
a = a + "!"
f'{a = }, {b = }, {a is b = }'   # a moved on, b didn't follow
```

#### `del` and Rebinding
* `del` removes a binding, not necessarily the object

```python
x = 5
y = x
del x
f'{y = }'
try: 
  f'{x}'
except NameError as e:
  f'x: {e = }'
```

## Address, Pointer, Reference

#### Address

```python


x = c_int(42)       
f'{x.value = }'            

addr = addressof(x)
f'{addr = }'
```

#### Pointer

```python
p = pointer(x)
f'{p.contents.value = }'

p.contents.value = 99
f'{x.value = }'            
```

#### `ctypes.POINTER`

```python
IntPtr = POINTER(c_int)
f'{IntPtr =}'
f'{type(p) == IntPtr = }'
f'{p.contents.value = }'
```

#### Null Pointer

```python
null_ptr = IntPtr()
f'IntPtr(): an instance — NULL pointer'
f'---------------------------------------'
f'{null_ptr}'
f'{bool(null_ptr) = }'
```

#### ctypes.cast

```python
f'ctypes.cast(addr, IntPtr)=  an instance — pointer aimed at addr which stores int to which x points'
f'---------------------------------------'
live_ptr = cast(addr, IntPtr)
f'{bool(live_ptr) = }'
f"{live_ptr.contents.value = }"
```

#### `ctypes.cast` part 2

```python
p1 = cast(addr, POINTER(c_int))
p2 = cast(p1, c_void_p)
p3 = cast(p2, POINTER(c_int))

f'{addr = }'
f'pointer to int: {p1 =}'
f'generic pointer no type: {p2 =}'
f'generic pointer to int pointer: {p3 =}'
f'p1, p2, p3 all hold the SAME address -- only the label changed each time'
```
