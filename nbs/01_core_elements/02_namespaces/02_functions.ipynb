# Functions

#### Parameters and Arguments

* `rank` in `def is_face(rank)` is the parameter.

```python
def is_face(rank):
    return rank > 10
```

* `face` in `is_face(face)` is the argument. 
* `7` is an argument too.

```python
face = 12
answer = is_face(face)
f'{face = }, {answer = }'
f'{is_face(7) = }'
```

* `return` gives `answer` its value. `answer` is `True`.
* `rank` exists while `is_face` runs. After the return, that name is gone. `face` and `answer` remain.

```python
try:
    f'{rank = }'
except NameError as e:
    f'{e = }'
```

* No arguments passed raises an Exception

```python
try:
    is_face()
except Exception as e:
    f'{e = }'
```

#### Call by Value

```python
def swap1(p, q):
    t = p 
    p = q
    q = t

x1 = c_int(1)
y1 = c_int(2)
f'{x1 = }, {y1 = }'
swap1(x1.value, y1.value)
f'{x1 = }, {y1 = }'
```

#### Call By Reference

```python
def swap2(p, q):
    t = p.contents.value
    p.contents.value = q.contents.value
    q.contents.value = t

x2 = c_int(1)
y2 = c_int(2)
x2_ptr = pointer(x2)
y2_ptr = pointer(y2)
f'{x2 = }, {y2 = }'
swap2(x2_ptr, y2_ptr)
f'{x2 = }, {y2 = }'
```

```python
def label(rank, suit):
    return f'{rank} of {suit}'

f'{label(12, "spades") = }'
f'{label(7, "hearts") = }'
```
