## Collections - Intro

* A collection holds values.
* Sequence — by position.
* Mapping — by name.

{{< include /includes/deck-of-cards.md >}}

#### array
* homogeneous — one typecode
* sequence
* mutable

```python
ranks = array('i', [1, 7, 12])
f'{ranks = }, {ranks.typecode = }'
f'{isinstance(ranks,Sequence) = }'
f'{issubclass(array,Sequence) = }'
```

#### `str`
* homogenous
* sequence
* immutable 

```python
f'{isinstance(c_char(65),Sequence) = }'
f'{isinstance("python",Sequence) = }'
f'{issubclass(str,Sequence) = }'
```

#### tuple
* homogeneous
* sequence
* immutable

```python
ranks_t = (1, 7, 12)
f'{ranks_t = }, {type(ranks_t) = }'
f'{isinstance(ranks_t,Sequence) = }'
f'{issubclass(tuple,Sequence) = }'
```

#### list
* heterogeneous
* sequence
* mutable

```python
card_l = [12, "spades"]
f'{card_l = }, {type(card_l) = }'
f'{isinstance(card_l,Sequence) = }'
f'{issubclass(list,Sequence) = }'
```

#### dict
* mapping
* mutable

```python
card_d = {"rank": 12, "suit": "spades"}
f'{card_d = }, {type(card_d) = }'
f'{card_d["rank"] = }'
f'{isinstance(card_d,Sequence) = }'
f'{issubclass(dict,Sequence) = }'
```

### Namespace
* A function's names are a dict.
* `SimpleNamespace` is a dict read with a dot.
* A dict is a namespace too.

```python
def label(rank, suit):
    return locals()

f'{label(12, "spades") = }'
```

```python
card_sn = SimpleNamespace(rank=12, suit="spades")
f'{card_sn.rank = }, {type(card_sn) = }'
```
