#### NOT - `~`
* `~` unary ~ (invert operator) is used to perform the bit complement or `not` operation.
*  we can't represent a single bit in python but now we can perform it on integers.


```python
x = c_uint8(0b01010101)
y = c_uint8(~x.value)
z = c_int8(~x.value)

f'{x.value = },  {x.value = :08b}'
f'{y.value = }, {y.value = :08b}'
f'{z.value = }, {z.value = :08b}'
```

### Byte Ordering
* Endianess
  * Little Endian - Most CPUs today
    * Most significant byte(MSB) goes last
  * Big Endian - TCP/IP protocols
    * Most significant byte(MSB) goes first
> Timelines
* Big Endian - IBM 360(1964)
* Little Endian - PDP-11(1970), Intel 8008(1972), Intel 8080/8086/x86(1974-1978),

https://en.wikipedia.org/wiki/Endianness

#### `to_bytes` - byte order

```python
a = 16784300
b = -16784300

a_le = a.to_bytes(length=4,byteorder='little',signed=True)
a_be = a.to_bytes(length=4,byteorder='big',signed=True)
b_le = b.to_bytes(length=4,byteorder='little',signed=True)
b_be = b.to_bytes(length=4,byteorder='big',signed=True)
f'{a_le = }, {a_le.hex() = }'
f'{a_be = }, {a_be.hex() = }'
f'{b_le = }, {b_le.hex() = }'
f'{b_be = }, {b_be.hex() = }'
```
