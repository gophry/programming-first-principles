#### Above/Below Scores


```python
#| exports
def above_35(score):
    """Return whether score is 35 or higher."""
    return score >= 35

def below_35(score):
    """Return whether score is 35 or lower."""
    return score <= 35
```

#### Even / Odd

```python
#| exports
def is_even(x):
    """Return whether x is divisible by 2."""
    return x % 2 == 0

def is_odd(x):
    """Return whether x is not divisible by 2."""
    return x % 2 != 0
```

#### Sums to

```python
#| exports
def sums_to(*values, T):
    """Return whether the sum of values equals T."""
    return sum(values) == T
```

```python
#| exports
from programming_first_principles.predicates import sums_to
def amounts_sum_to(*txns, T):
    """Return whether these are different transactions whose amounts sum to T."""
    amounts = [t.amount for t in txns]
    return len({t.id for t in txns}) == len(txns) and sums_to(*amounts, T=T)
```
